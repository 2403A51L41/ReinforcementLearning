```python
import random

# ==========================================
# MODIFIED EPSILON-GREEDY MULTI-ARMED BANDIT
# ==========================================

# Part A: Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Exploration probability
epsilon = 0.2

# Given reward patterns
rewards = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# ==========================================
# Part B: Initialize N, R and Q Tables
# ==========================================

# Number of times each machine is played
N = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Total reward of each machine
R = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Estimated reward (Q-value)
Q = {
    'A': 0.0,
    'B': 0.0,
    'C': 0.0
}

# Track which reward should be used next
reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Total reward received by player
total_reward = 0


# ==========================================
# Initial Table
# ==========================================

print("=" * 55)
print("MODIFIED EPSILON-GREEDY MULTI-ARMED BANDIT")
print("=" * 55)

print("\nInitial Table")
print("-" * 55)
print("Machine\tN (Count)\tR (Total)\tQ (Average)")

for machine in machines:
    print(f"{machine}\t{N[machine]}\t\t{R[machine]}\t\t{Q[machine]:.2f}")


# ==========================================
# Perform 10 Plays
# ==========================================

for play in range(1, N_play + 1):

    print("\n" + "=" * 55)
    print(f"PLAY {play}")
    print("=" * 55)

    # Step 1: Generate random number
    r = random.random()

    # Find the highest Q-value
    highest_Q_value = max(Q.values())

    # Find all machines with the highest Q-value
    highest_Q_machines = [
        machine for machine in machines
        if Q[machine] == highest_Q_value
    ]

    # Randomly choose if there is a tie
    highest_Q_machine = random.choice(highest_Q_machines)

    print(f"Random Number = {r:.2f}")
    print(f"Highest Q Machine = {highest_Q_machine}")


    # ==========================================
    # Part C: Modified Epsilon-Greedy Decision
    # ==========================================

    if r < epsilon:

        print("Decision = EXPLORATION")

        # Select a machine except the highest Q machine
        available_machines = [
            machine for machine in machines
            if machine != highest_Q_machine
        ]

        selected_machine = random.choice(available_machines)

    else:

        print("Decision = EXPLOITATION")

        # Select the highest Q-value machine
        selected_machine = highest_Q_machine


    # ==========================================
    # Part D: Automatic Reward Collection
    # ==========================================

    # Get next reward automatically
    # If all 4 rewards are used, start again from the beginning
    index = reward_index[selected_machine] % len(rewards[selected_machine])

    reward = rewards[selected_machine][index]

    # Move to the next reward
    reward_index[selected_machine] += 1

    print(f"Selected Machine = {selected_machine}")
    print(f"Reward Obtained = {reward}")


    # ==========================================
    # Part E: Update N, R and Q
    # ==========================================

    # Update participation count
    N[selected_machine] += 1

    # Update total reward
    R[selected_machine] += reward

    # Incremental sample-average Q-value update
    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )

    # Update player's total reward
    total_reward += reward


    # ==========================================
    # Display Table After Every Play
    # ==========================================

    print("\nTable After Play", play)
    print("-" * 55)
    print("Machine\tN (Count)\tR (Total)\tQ (Average)")

    for machine in machines:
        print(
            f"{machine}\t{N[machine]}\t\t"
            f"{R[machine]}\t\t{Q[machine]:.2f}"
        )

    print(f"\nTotal Player Reward = {total_reward}")


# ==========================================
# Part F: Final Results
# ==========================================

print("\n" + "=" * 55)
print("FINAL RESULTS AFTER 10 PLAYS")
print("=" * 55)

print("\nMachine\tParticipation(N)\tTotal Reward(R)\tQ-Value")

for machine in machines:
    print(
        f"{machine}\t{N[machine]}\t\t\t"
        f"{R[machine]}\t\t\t{Q[machine]:.2f}"
    )

print("\nTotal Reward Received by Player =", total_reward)

# Find the best machine
best_machine = max(Q, key=Q.get)

print("Best Machine =", best_machine)
print("Highest Q-Value =", round(Q[best_machine], 2))
```
